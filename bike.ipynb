```python
# Fill categorical columns with their most frequent value
categorical_columns = ["season", "holiday", "workingday", "weathersit"]
for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

# Fill numerical columns with their median
numeric_columns = ["yr", "mnth", "temp", "atemp", "hum","windspeed", "casual", "registered"]
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df[col] = df[col].fillna(df[col].median())
print(df.isnull().sum())
print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
numerical_columns = df.select_dtypes(include=np.number).columns

# Detect potential outliers using IQR
for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outlier_count = (
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ).sum()

    print(f"{col}: {outlier_count} potential outliers")
```

```python
for col in df.select_dtypes(include=['int','float']).columns:
    sns.boxplot(df[col])
    plt.title(f'Box Plot of {col}')
    plt.show()
```

```python
# Outlier Capping (excluding the target column 'cnt')
for i in df.select_dtypes(include=['int', 'float']).columns:
    if i in ['cnt', 'casual', 'registered']:
     continue

    q1 = df[i].quantile(0.25)
    q3 = df[i].quantile(0.75)
    IQR = q3 - q1

    lower = q1 - 1.5 * IQR
    upper = q3 + 1.5 * IQR

    df[i] = df[i].clip(lower, upper)
```

### 3.feature engineering

```python
#weekend feature
#This feature identifies whether a rental record falls on a weekend
df['is_weekend'] = df['weekday'].isin([0, 6]).astype(int)
df[['weekday', 'is_weekend']].head()
```

```python
#time-of-day feature
#group the 24 hours into four periods.
def get_time_period(hour):
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'

df['time_period'] = df['hr'].apply(get_time_period)

df[['hr', 'time_period']].head(24)
```

```python
#rush-hour feature
df['is_rush_hour'] = df['hr'].isin([7, 8, 9, 17, 18, 19]).astype(int)

df[['hr', 'is_rush_hour']].head(10)
```

```python
#Extract the month and year from the date
#dataset already has mnth and yr, but we can create date-derived features for consistency.
df['month_from_date'] = df['dteday'].dt.month
df['year_from_date'] = df['dteday'].dt.year
df[['dteday', 'month_from_date', 'year_from_date']].head()
```

```python
#new features
new_features = [
    'is_weekend',
    'time_period',
    'is_rush_hour',
    'month_from_date',
    'year_from_date'
]

df[new_features].head()
```

```python
xgb_random = RandomizedSearchCV(
    estimator=XGBRegressor(
        random_state=42,
        objective='reg:squarederror',
        n_jobs=1
    ),
    param_distributions=xgb_param_grid,
    n_iter=20,
    cv=5,
    scoring='neg_mean_squared_error',
    random_state=42,
    n_jobs=-1,
    verbose=2
)

xgb_random.fit(X_train, y_train)
```

```python
print("Best XGBoost Parameters:")
print(xgb_random.best_params_)

print("\nBest Cross-Validation RMSE:")
print(np.sqrt(-xgb_random.best_score_))
```

```python
xgb_random_model = xgb_random.best_estimator_

xgb_random_predictions = xgb_random_model.predict(X_test)

xgb_random_mae = mean_absolute_error(y_test, xgb_random_predictions)

xgb_random_rmse = np.sqrt(
    mean_squared_error(y_test, xgb_random_predictions)
)

xgb_random_r2 = r2_score(y_test, xgb_random_predictions)

print("Randomized Search - XGBoost Performance")
print("MAE:", xgb_random_mae)
print("RMSE:", xgb_random_rmse)
print("R2 Score:", xgb_random_r2)
```

```python
from sklearn.svm import SVR
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
svr_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('svr', SVR())
])
```

```python
svr_param_grid = {
    'svr__kernel': ['rbf', 'linear', 'poly'],
    'svr__C': [0.1, 1, 10, 50, 100, 200],
    'svr__epsilon': [0.01, 0.1, 0.2, 0.5, 1],
    'svr__gamma': ['scale', 'auto']
}
```

```python
svr_random = RandomizedSearchCV(
    estimator=svr_pipeline,
    param_distributions=svr_param_grid,
    n_iter=20,
    cv=5,
    scoring='neg_mean_squared_error',
    random_state=42,
    n_jobs=-1,
    verbose=2
)

svr_random.fit(X_train, y_train)
```

```python
print("Best SVR Parameters:")
print(svr_random.best_params_)

print("\nBest Cross-Validation RMSE:")
print(np.sqrt(-svr_random.best_score_))
```

```python
svr_random_model = svr_random.best_estimator_

svr_random_predictions = svr_random_model.predict(X_test)

svr_random_mae = mean_absolute_error(y_test, svr_random_predictions)

svr_random_rmse = np.sqrt(
    mean_squared_error(y_test, svr_random_predictions)
)

svr_random_r2 = r2_score(y_test, svr_random_predictions)

print("Randomized Search - SVR Performance")
print("MAE:", svr_random_mae)
print("RMSE:", svr_random_rmse)
print("R2 Score:", svr_random_r2)
```
